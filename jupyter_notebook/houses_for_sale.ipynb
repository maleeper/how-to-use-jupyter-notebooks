```python
# There are over 600 agents!We will use agent logo number as the agent ID
# Get the number between the brackets in the URL
# print the full agent logo URLs
#print(df['agent_logo'].unique())
unique_agents = df['agent_logo'].unique()
# Check unique values in 'agent_logo' column
print(f"The agent_logo has {len(unique_agents)} unique values")
# Display the unique agent logos
for agent in unique_agents:
    print(agent)

# Extract the agent ID from the agent_logo URL replace the agent_logo column with the agent_id
# We have some NaN values, so we will fill them with '0' and convert to int

df['agent_id'] = df['agent_logo'].str.extract(r'(\d+)', expand=False).fillna('0').astype(int)
unique_agentIDs = df['agent_id'].unique()
# Check unique values in 'agent_id' column
print(f"The agent_id has {len(unique_agentIDs)} unique values")
# Display the unique agent IDs
for agent_id in unique_agentIDs:
    print(agent_id)
```

```python
# Let's get the first part of the postcode. It is the last part of the address eg. B1
# Notice that some postcodes are not in the format we expect, so we will need to remove them

df['postcode'] = df['address'].str.rstrip(',').str.split().str[-1].str.extract(r'\b([A-Z]{1,2}\d[A-Z\d]?)\b', expand=False)

# Check the new 'postcode' column

unique_postcodes = df['postcode'].unique()
print(f"The postcode has {len(unique_postcodes)} unique values")
# Display the unique postcodes
for postcodes in unique_postcodes:
    print(postcodes)
# number of rows with NaN postcodes
nan_postcodes = df['postcode'].isna().sum()
print(f"There are {nan_postcodes} rows with NaN postcodes")
# We can now drop the rows with NaN postcodes
df = df.dropna(subset=['postcode'])
unique_postcodes = df['postcode'].unique()
print(f"The postcode has {len(unique_postcodes)} unique values")
```

```python
# convert the 'price' column to numeric, errors='coerce' will convert non-numeric values to NaN
df['price'] = pd.to_numeric(df['price'].str.replace('£', '').str.replace(',', ''), errors='coerce')

# convert the agent_id column to int
df['agent_id'] = df['agent_id'].astype(int)
print(df.dtypes)
```

```python
# Need to split the 'rooms_and_sqft' column into separate columns for beds, baths, receptions, and sqft.
# # Regex with optional receptions and optional sqft
pattern = (
    r'(?P<beds>\d+)\s*beds?'              # beds
    r'(?P<baths>\d+)\s*baths?'            # baths
    r'(?:'                                # optional group for receptions
    r'(?P<receptions>\d+)\s*receptions?'  
    r')?'                                 # end optional
    r'(?:\s*(?P<sqft>\d+)\s*sq)?'         # optional sqft
)

# Extract and convert to nullable integers
extracted = df['rooms_and_sqft'].str.extract(pattern)
for col in ['beds', 'baths', 'receptions', 'sqft']:
    df[col] = pd.to_numeric(extracted[col], errors='coerce').astype('Int64')

# Now we can drop the 'rooms_and_sqft' column
df = df.drop(columns=['rooms_and_sqft'])
# Show the first few rows to confirm the changes
df.head(20)
```

```python
# 1) read cleaned file
df_features = pd.read_csv("key_points_cleaned.csv")

# 2) split mixed-case strings → individual tokens
tokens = (
    df_features['key_points']
        .fillna('')
        .str.split('|')
        .explode()
        .str.strip()
)
tokens = tokens[tokens.ne('')]        # drop empties

# 3) count token frequencies
counts = tokens.value_counts()
print(counts.head(15))                # inspect the top 15

# pick the most common features
top_n   = 5
keepers = counts.head(top_n).index

# 4) add Boolean “is_…” columns for those features
for feat in keepers:
    col_name = 'is_' + re.sub(r'\s+', '_', feat.lower())
    df_features[col_name] = df_features['key_points']\
        .str.contains(feat, case=False, na=False)

# preview
print(df_features.head(20))

# it must be either Freehold or Leasehold check the is_freehold and is_leasehold columns
# 1️⃣ if both are False → make Freehold True
mask_both_false = (~df_features['is_freehold']) & (~df_features['is_leasehold'])
df_features.loc[mask_both_false, 'is_freehold'] = True

# 2️⃣ if both are True  → turn Freehold off (so exactly one flag stays True)
mask_both_true = df_features['is_freehold'] & df_features['is_leasehold']
df_features.loc[mask_both_true, 'is_freehold'] = False
df_features[['is_freehold', 'is_leasehold']].head(10)


```
